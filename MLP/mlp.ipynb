# Multi Layer Perceptron (MLP)

A **Multi Layer Perceptron (MLP)** is a feedforward neural network used for classification and regression tasks. It consists of an input layer, one or more hidden layers, and an output layer.

## Core Concept

Each neuron receives inputs, applies weights and a bias, and passes the result through an activation function.

`output = activation(XW + b)`

Hidden layers allow the model to learn nonlinear patterns that a single-layer perceptron cannot represent.

## Training Workflow

1. **Forward pass**: Generate predictions from input data.
2. **Loss calculation**: Measure prediction error.
3. **Backpropagation**: Compute gradients for each parameter.
4. **Optimization**: Update weights and biases to reduce loss.

## Minimal NumPy Example

```python
import numpy as np

np.random.seed(42)

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
y = np.array([[0], [1], [1], [0]], dtype=float)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_grad(a):
    return a * (1 - a)

W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))
W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

for _ in range(10000):
    h = sigmoid(X @ W1 + b1)
    y_pred = sigmoid(h @ W2 + b2)

    d_out = (y_pred - y) * sigmoid_grad(y_pred)
    d_hidden = (d_out @ W2.T) * sigmoid_grad(h)

    W2 -= 0.5 * h.T @ d_out
    b2 -= 0.5 * d_out.sum(axis=0, keepdims=True)
    W1 -= 0.5 * X.T @ d_hidden
    b1 -= 0.5 * d_hidden.sum(axis=0, keepdims=True)

print((y_pred > 0.5).astype(int))
```

## Summary

MLPs are foundational neural networks. They combine layered transformations, nonlinear activations, and gradient-based learning to model complex relationships in data.
