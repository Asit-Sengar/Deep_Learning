# Multi Layer Perceptron (MLP)

A **Multi Layer Perceptron (MLP)** is a feedforward artificial neural network made of layers of neurons. It is one of the most common starting points for learning deep learning because it shows the core ideas behind neural networks: weights, biases, activation functions, forward propagation, loss, and backpropagation.

## What Makes It "Multi Layer"?

A single perceptron has only one layer of learnable weights and can solve only simple linear problems. An MLP has at least one **hidden layer** between the input and output layers.

Typical structure:

1. **Input layer**: Receives the feature values.
2. **Hidden layer(s)**: Learn intermediate patterns by applying weights, biases, and nonlinear activation functions.
3. **Output layer**: Produces the final prediction, such as a class label or numeric value.

## How an MLP Works

Each neuron computes a weighted sum of its inputs, adds a bias, and passes the result through an activation function.

For one layer:

`z = XW + b`

`a = activation(z)`

Where:

- `X` is the input data.
- `W` contains the weights.
- `b` contains the biases.
- `z` is the linear result before activation.
- `a` is the activated output passed to the next layer.

Activation functions such as **ReLU**, **sigmoid**, or **tanh** add nonlinearity. Without nonlinear activations, stacking many layers would still behave like one linear model.

## Training Process

Training an MLP means adjusting weights and biases so predictions become more accurate.

The usual training loop is:

1. **Forward pass**: Send input data through the network to get predictions.
2. **Loss calculation**: Compare predictions with true labels using a loss function.
3. **Backpropagation**: Compute how much each weight contributed to the error.
4. **Optimization**: Update weights using an optimizer such as gradient descent.

Over many epochs, the network learns useful internal representations of the data.

## Tiny NumPy Example

The code below builds a small MLP from scratch to learn the XOR pattern. XOR is useful because it is not linearly separable, so a network needs a hidden layer to solve it.

```python
import numpy as np

np.random.seed(42)

# XOR dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

y = np.array([[0], [1], [1], [0]], dtype=float)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(output):
    return output * (1 - output)

# Network: 2 inputs -> 4 hidden neurons -> 1 output
W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))
W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

learning_rate = 0.5
epochs = 10000

for epoch in range(epochs):
    # Forward pass
    hidden_input = X @ W1 + b1
    hidden_output = sigmoid(hidden_input)
    final_input = hidden_output @ W2 + b2
    prediction = sigmoid(final_input)

    # Mean squared error derivative
    error = prediction - y

    # Backpropagation
    output_delta = error * sigmoid_derivative(prediction)
    hidden_delta = (output_delta @ W2.T) * sigmoid_derivative(hidden_output)

    # Gradient descent updates
    W2 -= learning_rate * (hidden_output.T @ output_delta)
    b2 -= learning_rate * output_delta.sum(axis=0, keepdims=True)
    W1 -= learning_rate * (X.T @ hidden_delta)
    b1 -= learning_rate * hidden_delta.sum(axis=0, keepdims=True)

print("Predictions after training:")
print(np.round(prediction, 3))
print("\nRounded class predictions:")
print((prediction > 0.5).astype(int))
```

## Key Terms

- **Weight**: A learnable value that controls the strength of a connection between neurons.
- **Bias**: A learnable offset that helps shift the activation of a neuron.
- **Activation function**: A function that introduces nonlinearity into the model.
- **Loss function**: A measure of how wrong the model's predictions are.
- **Gradient descent**: An optimization method that updates parameters in the direction that reduces loss.
- **Backpropagation**: The algorithm used to calculate gradients throughout the network.

## Why MLPs Matter

MLPs can approximate complex nonlinear relationships, making them useful for classification, regression, and pattern recognition tasks. They are also the foundation for many deeper neural network architectures. Even though modern deep learning often uses CNNs, RNNs, and Transformers, the basic ideas inside an MLP still appear everywhere.
