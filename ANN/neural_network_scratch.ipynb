```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../datasets/placement.csv")
```

```python
#it shows random rows
df.sample(5, random_state = 42)
```

```python
#we could have also declared the sigmoid function to introduce non linearity
def sigmoid(x):
    return 1/(1 + math.exp(-1* x))
```

```python
def step_function(x):
    if x<0.5:
        return 0
    elif x > 0.5:
        return 1
```

```python
#initialising all weights and biases to 0
#we can also declare them with random values
rng = np.random.default_rng()

weight1 = rng.integers(1,101) 
weight2 =  rng.integers(1,101) 
weight3 =  rng.integers(1,101) 
weight4 =  rng.integers(1,101) 
weight5 =  rng.integers(1,101) 
weight6 =  rng.integers(1,101) 
b1 =  rng.integers(1,101) 
b2 =  rng.integers(1,101) 
b3 =  rng.integers(1,101) 

learning_rate = 0.01


for i in range(1000):#outer loop
    error = 0#error declaration
    for i in range(df.shape[0]):
        #forward propagation
        output1 = sigmoid(weight1 * df.iloc[i,2] + weight3 * df.iloc[i,3] + b1)
        output2 = sigmoid(weight2 * df.iloc[i,2] + weight4 * df.iloc[i,3] + b2)


        output3 = sigmoid(weight5*output1 + weight6*output2 + b3)

        prediction = step_function(output3)

        #now we should calculate loss

        error = error + (prediction - df.iloc[i][4])

        
    print("loss is:" , error)

    #now the backpropagation part

    # Calculate and accumulate gradients for every training example.
    dw1 = dw2 = dw3 = dw4 = dw5 = dw6 = 0.0
    db1 = db2 = db3 = 0.0

    for j in range(df.shape[0]):
        x1 = df.iloc[j, 2]
        x2 = df.iloc[j, 3]
        target = df.iloc[j, 4]

        # Forward values needed to apply the chain rule.
        hidden1 = sigmoid(weight1 * x1 + weight3 * x2 + b1)
        hidden2 = sigmoid(weight2 * x1 + weight4 * x2 + b2)
        output = sigmoid(weight5 * hidden1 + weight6 * hidden2 + b3)

        # Output neuron gradient for half squared error.
        delta_output = (output - target) * output * (1 - output)
        dw5 += delta_output * hidden1
        dw6 += delta_output * hidden2
        db3 += delta_output

        # Propagate the error back to the hidden neurons.
        delta_hidden1 = delta_output * weight5 * hidden1 * (1 - hidden1)
        delta_hidden2 = delta_output * weight6 * hidden2 * (1 - hidden2)
        dw1 += delta_hidden1 * x1
        dw3 += delta_hidden1 * x2
        db1 += delta_hidden1
        dw2 += delta_hidden2 * x1
        dw4 += delta_hidden2 * x2
        db2 += delta_hidden2

    # Update weights and biases using the mean gradient.
    sample_count = df.shape[0]
    weight1 -= learning_rate * dw1 / sample_count
    weight2 -= learning_rate * dw2 / sample_count
    weight3 -= learning_rate * dw3 / sample_count
    weight4 -= learning_rate * dw4 / sample_count
    weight5 -= learning_rate * dw5 / sample_count
    weight6 -= learning_rate * dw6 / sample_count
    b1 -= learning_rate * db1 / sample_count
    b2 -= learning_rate * db2 / sample_count
    b3 -= learning_rate * db3 / sample_count
    
    

        



```
