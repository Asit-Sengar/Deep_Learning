### Step 7: Train the perceptron model
This cell initializes weights and repeatedly updates them using prediction errors to learn a binary decision boundary for placement.

```python
weight1 = 0
weight2 = 0
bias = 0

def step_function(z):
    if z >=0:
        return 1
    else:
        return 0

learning_rate = 0.01

for epoch in range(1000):

    prediction_array = []
    for i in range(df.shape[0]):#this loop is for taking values of features of every student
        z = weight1 * df.iloc[i ,1] + weight2  * df.iloc[i,2] + bias
    
        #prediction using the old weight values
        prediction = step_function(z)#this shall return either 0 or 1
        
        #calculation of new weights
        error = df.iloc[i , 3] - prediction

        weight1 = weight1  + learning_rate *  error * df.iloc[i , 1]#for cgpa 
        weight2 = weight2  + learning_rate *  error * df.iloc[i , 2]#for iq
        bias = bias + learning_rate * error


        #accuracy calculation and displaying it
        prediction_array.append(prediction)#storing all the predictions

    correct_answers = 0
    for i in range(len(prediction_array)):
        if(prediction_array[i]==df.iloc[i , 3]):
            correct_answers = correct_answers + 1

    print("accuracy is:" , (correct_answers/df.shape[0]) * 100)
```
